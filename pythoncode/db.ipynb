```python
import sqlite3
import pandas as pd
import os
from dotenv import load_dotenv
load_dotenv()

dbpath:str = os.getenv("SQLITE_FILE","")

content:str = ""
with open("../database/init.sql","r",encoding="utf-8") as file: content = file.read()

conn = sqlite3.connect(dbpath)
cur = conn.cursor()

cur.executescript(content)
conn.commit()
```
